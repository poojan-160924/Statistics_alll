```python
#Task:1
# Matrix representing Instagram likes
likes = [
    [120, 15],
    [200, 25]
]

# Matrix representing Instagram comments
comments = [
    [2, 1],
    [3, 4]
]

# Multiply the two matrices
result = [
    [0, 0],
    [0, 0]
]

for i in range(2):
    for j in range(2):
        for k in range(2):
            result[i][j] += likes[i][k] * comments[k][j]

# Display the matrices
print("Likes Matrix:")
for row in likes:
    print(row)

print("\nComments Matrix:")
for row in comments:
    print(row)

print("\nResult Matrix:")
for row in result:
    print(row)
```

```python
# Task:2
# Function to predict output using linear regression

def linear_regression_predict(x, m, c):

    y = m * x + c
    return y

# Given values
x = 5
m = 4
c = 10

# Calculate predicted output
result = linear_regression_predict(x, m, c)

print("Input value:", x)
print("Slope:", m)
print("Intercept:", c)
print("Predicted output:", result)
```

```python
# Task:3
# Function to calculate delivery time

def linear_regression_predict(x, m, c):
    return m * x + c

# Given values
m = 4
c = 10

# Delivery distances in kilometers
distances = [2, 5, 8]

print("Zomato Delivery Time Prediction")
print("--------------------------------")

for distance in distances:

    delivery_time = linear_regression_predict(distance, m, c)

    print("Distance:", distance, "km")
    print("Predicted delivery time:", delivery_time, "minutes")
    print()
```
