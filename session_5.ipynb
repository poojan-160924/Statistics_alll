```python
# Task:2
import random

# Create a list of 1000 user IDs
users = []

for i in range(1, 1001):
    users.append("user" + str(i))

# Select 50 random users
selected_users = random.sample(users, 50)

# Print the selected users
print("Total number of users:", len(users))
print("\nRandomly selected 50 users:")

for user in selected_users:
    print(user)

print("\nTotal selected users:", len(selected_users))
```

```python
# Task:4
"""Question: A survey about music preferences is only sent to users who recently streamed Bollywood songs on Spotify. Identify and explain the type of bias this introduces in the sample.

Answer:

This situation introduces Selection Bias because the survey is sent only to users who recently listened to Bollywood songs.
Users who listen to English, Punjabi, Gujarati, or other music genres are not included in the survey. As a result, Bollywood music may appear more popular than it actually is among all Spotify users.

How to avoid this bias:
Spotify should randomly select users from different music preferences, listening habits, and age groups. 
This would make the survey more balanced and representative."""
```
