```python
#Task:1
import numpy as np
import matplotlib.pyplot as plt

# Generate 100 random numbers
data = np.random.normal(50, 10, 100)

# Calculate mean and standard deviation
mean = np.mean(data)
std = np.std(data)

print("Mean:", round(mean, 2))
print("Standard Deviation:", round(std, 2))

# Plot histogram
plt.hist(data, bins=15, density=True, edgecolor="black")

# Plot bell curve
x = np.linspace(20, 80, 100)

y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(
    -0.5 * ((x - mean) / std) ** 2
)

plt.plot(x, y)

plt.title("Normal Distribution Bell Curve")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

```python
# Task:2
import numpy as np
from scipy import stats

# Daily step counts for 20 days
steps = [
    4000, 4500, 5000, 5000, 5500,
    5500, 6000, 6000, 6000, 6500,
    6500, 6500, 7000, 7000, 7500,
    7500, 8000, 8000, 8500, 8500
]

# Calculate mean, median and mode
mean = np.mean(steps)
median = np.median(steps)
mode = stats.mode(steps, keepdims=True).mode[0]

print("Daily steps:", steps)
print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)

# Compare the values
if abs(mean - median) < 500:
    print("The data is approximately normally distributed.")
else:
    print("The data may not be normally distributed.")
```

```python
# Task:3
import numpy as np

# Generate scores of 200 players
np.random.seed(42)

scores = np.random.normal(70, 15, 200)

# Calculate mean and standard deviation
mean_score = np.mean(scores)
std_score = np.std(scores)

# Find scores within one standard deviation
lower_limit = mean_score - std_score
upper_limit = mean_score + std_score

count = 0

for score in scores:
    if lower_limit <= score <= upper_limit:
        count += 1

print("Total players:", len(scores))
print("Average score:", round(mean_score, 2))
print("Standard deviation:", round(std_score, 2))

print("Lower limit:", round(lower_limit, 2))
print("Upper limit:", round(upper_limit, 2))

print("Players within one standard deviation:", count)
```

```python
# Task:4
import numpy as np
import matplotlib.pyplot as plt

# Delivery times of 100 Zomato orders
np.random.seed(42)

delivery_times = np.random.normal(30, 5, 100)

# Calculate mean and standard deviation
mean_time = np.mean(delivery_times)
std_time = np.std(delivery_times)

print("Average delivery time:", round(mean_time, 2), "minutes")
print("Standard deviation:", round(std_time, 2), "minutes")

# Create histogram
plt.hist(delivery_times, bins=12, edgecolor="black", density=True)

plt.title("Zomato Delivery Time Distribution")
plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Frequency")

plt.show()

print("Most orders are delivered close to the average delivery time.")
```
