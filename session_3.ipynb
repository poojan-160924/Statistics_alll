```python
# Task : 1

steps = [4500, 7000, 5000, 8000, 4000, 9000, 6000]

#1: Calculate the mean
mean_steps = sum(steps) / len(steps)

print("Mean:", round(mean_steps, 2))

#2: Calculate squared differences
squared_differences = []

for step in steps:
    difference = step - mean_steps
    squared = difference ** 2
    squared_differences.append(squared)

    print(step, "->", round(squared, 2))

#3: Calculate variance
variance = sum(squared_differences) / len(steps)

print("Variance:", round(variance, 2))
```

```python
# Task:2

import statistics

def calculate_standard_deviation(scores):
    standard_deviation = statistics.pstdev(scores)
    return round(standard_deviation, 2)

scores = [78, 85, 90, 88, 76]
result = calculate_standard_deviation(scores)

print("Exam scores:", scores)
print("Standard deviation:", result)
```

```python
# Task:3
import statistics

# Daily spending of two friends
friend_A = [200, 200, 200, 200, 200]
friend_B = [100, 300, 150, 400, 50]

# Calculate standard deviation
sd_A = statistics.pstdev(friend_A)
sd_B = statistics.pstdev(friend_B)

print("Friend A standard deviation:", round(sd_A, 2))
print("Friend B standard deviation:", round(sd_B, 2))

# Compare spending habits
if sd_A < sd_B:
    print("Friend A has more consistent spending.")
else:
    print("Friend B has more consistent spending.")
```

```python
# Task:4
import statistics

salaries = [28000, 29000, 27000, 60000, 26500, 27500]

# Calculate average salary
mean_salary = statistics.mean(salaries)

# Calculate variance and standard deviation
variance = statistics.pvariance(salaries)
standard_deviation = statistics.pstdev(salaries)

print("Employee salaries:", salaries)
print("Average salary:", mean_salary)
print("Variance:", round(variance, 2))
print("Standard deviation:", round(standard_deviation, 2))
```
