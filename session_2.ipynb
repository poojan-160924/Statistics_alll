```python
# Task:1

import statistics

daily_plays = [120, 135, 150, 200, 120, 90, 200]
mean_plays = statistics.mean(daily_plays)

print("Daily song plays:", daily_plays)
print("Mean number of plays:", mean_plays)
```

```python
# Task:2

import statistics

delivery_times = [30, 25, 40, 35, 30, 45, 30]
median_time = statistics.median(delivery_times)
print("Median delivery time:", median_time, "minutes")
```

```python
# Task:3

from statistics import mode

def most_common_rating(ratings):
    return mode(ratings)

ratings = [5, 4, 4, 3, 5, 4, 2, 4]
result = most_common_rating(ratings)
print("Most common rating:", result)
```

```python
# Task:4

import statistics

channel_A = [100, 120, 110, 105, 115]
channel_B = [100, 120, 110, 105, 5000]
channel_C = [80, 90, 90, 100, 110]

channels = {
    "Channel A": channel_A,
    "Channel B": channel_B,
    "Channel C": channel_C
}

for name, views in channels.items():
    print("\n", name)
    print("Mean:", statistics.mean(views))
    print("Median:", statistics.median(views))

    try:
        print("Mode:", statistics.mode(views))
    except statistics.StatisticsError:
        print("Mode: No mode")
```

```python
# Task:5

import pandas as pd

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"

df = pd.read_csv(url, sep=";")

print("First five rows:")
print(df.head())

alcohol_mean = df["alcohol"].mean()
alcohol_median = df["alcohol"].median()
alcohol_mode = df["alcohol"].mode()[0]

print("\nStatistics for alcohol column:")
print("Mean:", alcohol_mean)
print("Median:", alcohol_median)
print("Mode:", alcohol_mode)
```
