```python
# Task:1

# Average daily usage time in minutes

instagram_time = [60, 75, 80, 90, 65, 85, 70, 95, 100, 80]
youtube_time = [45, 50, 60, 55, 65, 40, 70, 55, 60, 50]

from scipy import stats

# Perform one-tailed independent t-test
t_stat, p_value = stats.ttest_ind(
    instagram_time,
    youtube_time,
    alternative="greater"
)

print("Instagram Average:", sum(instagram_time) / len(instagram_time))
print("YouTube Average:", sum(youtube_time) / len(youtube_time))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("Reject the null hypothesis.")
    print("People spend significantly more time on Instagram.")
else:
    print("Fail to reject the null hypothesis.")
```

```python
# Task:2
from scipy import stats

# Group A: New homepage
mean_A = 350
std_A = 50
n_A = 30

# Group B: Old homepage
mean_B = 320
std_B = 50
n_B = 30

# Perform two-sample t-test
t_stat, p_value = stats.ttest_ind_from_stats(
    mean1=mean_A,
    std1=std_A,
    nobs1=n_A,
    mean2=mean_B,
    std2=std_B,
    nobs2=n_B
)

print("T-statistic:", round(t_stat, 4))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

```python
# Task:3
from scipy import stats

# Group A: Old WhatsApp design
control_group = [30, 35, 40, 32, 38, 36, 34, 31, 39, 33]

# Group B: New dark mode design
variant_group = [40, 45, 42, 48, 44, 46, 41, 43, 47, 45]

# Calculate average usage time
control_mean = sum(control_group) / len(control_group)
variant_mean = sum(variant_group) / len(variant_group)

# Perform independent t-test
t_stat, p_value = stats.ttest_ind(
    control_group,
    variant_group
)

print("Control Group Average:", control_mean)
print("Variant Group Average:", variant_mean)
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("The new dark mode has a significant effect.")
else:
    print("There is no significant effect.")
```

```python
# Task:4
from scipy.stats import chi2_contingency

# Clicks and non-clicks
# Assuming 1000 users in each group

new_button = [520, 480]
old_button = [480, 520]

# Create a comparison table
data = [new_button, old_button]

# Perform chi-square test
chi2, p_value, dof, expected = chi2_contingency(data)

print("Chi-square value:", round(chi2, 4))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("The difference is statistically significant.")
else:
    print("The difference may be due to chance.")
```
