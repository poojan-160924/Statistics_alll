```python
#Task:1
# Function to calculate posterior probability

def bayes_posterior(prior, likelihood, evidence):

    posterior = (prior * likelihood) / evidence

    return posterior


# Given probabilities
prior = 0.4
likelihood = 0.7
evidence = 0.5

# Calculate posterior probability
result = bayes_posterior(prior, likelihood, evidence)

print("Prior Probability:", prior)
print("Likelihood:", likelihood)
print("Evidence:", evidence)

print("Posterior Probability:", round(result, 2))
print("Percentage:", round(result * 100, 2), "%")
```

```python
# Task:2

# Probability of a restaurant being premium
prior_premium = 0.10

# Probability of a discount from a premium restaurant
likelihood = 0.80

# Probability of a discount from a regular restaurant
regular_discount = 0.20

# Probability of a regular restaurant
prior_regular = 0.90

# Calculate total probability of getting a discount
evidence = (
    prior_premium * likelihood
    + prior_regular * regular_discount
)

# Apply Bayes theorem
posterior = (prior_premium * likelihood) / evidence

print("Prior Probability:", prior_premium)
print("Likelihood:", likelihood)
print("Evidence:", round(evidence, 2))
print("Probability of Premium Restaurant:", round(posterior, 4))
print("Percentage:", round(posterior * 100, 2), "%")
```

```python
# Task:3
# List of 20 email subjects with their categories

emails = [
    ("Free gift for you", "spam"),
    ("Win free shopping voucher", "spam"),
    ("Get free money now", "spam"),
    ("Free offer expires today", "spam"),
    ("Claim your free prize", "spam"),
    ("Congratulations, free reward", "spam"),
    ("Free mobile recharge", "spam"),
    ("Limited free deal", "spam"),
    ("Meeting scheduled for tomorrow", "not spam"),
    ("Project report submission", "not spam"),
    ("Your order has been shipped", "not spam"),
    ("College assignment details", "not spam"),
    ("Team meeting at 10 AM", "not spam"),
    ("Monthly bank statement", "not spam"),
    ("Your appointment is confirmed", "not spam"),
    ("Free time for a meeting", "not spam"),
    ("Important account update", "not spam"),
    ("Payment received successfully", "not spam"),
    ("Welcome to our newsletter", "not spam"),
    ("Your password has been changed", "not spam")
]

# Count total spam and non-spam emails
total_emails = len(emails)

spam_emails = [email for email, category in emails
               if category == "spam"]

normal_emails = [email for email, category in emails
                 if category == "not spam"]

# Count emails containing the word free
spam_free = sum("free" in email.lower()
                for email in spam_emails)

normal_free = sum("free" in email.lower()
                  for email in normal_emails)

# Calculate probabilities
prior_spam = len(spam_emails) / total_emails

prior_normal = len(normal_emails) / total_emails

likelihood = spam_free / len(spam_emails)

normal_likelihood = normal_free / len(normal_emails)

# Calculate probability of the word free
evidence = (
    prior_spam * likelihood
    + prior_normal * normal_likelihood
)

# Apply Bayes theorem
posterior = (prior_spam * likelihood) / evidence

print("Total emails:", total_emails)
print("Spam emails:", len(spam_emails))
print("Normal emails:", len(normal_emails))

print("Spam emails containing free:", spam_free)
print("Normal emails containing free:", normal_free)

print("Probability of spam:", round(posterior, 4))
print("Percentage:", round(posterior * 100, 2), "%")
```

```python
# Task:4
# Instagram fake account detection using Bayes theorem

# Probability that an account is fake
prior_fake = 0.05

# Probability of high activity from a fake account
likelihood = 0.80

# Probability of high activity from a genuine account
normal_activity = 0.10

# Probability that an account is genuine
prior_genuine = 0.95

# Calculate total probability of high activity
evidence = (
    prior_fake * likelihood
    + prior_genuine * normal_activity
)

# Apply Bayes theorem
posterior = (prior_fake * likelihood) / evidence

print("Prior probability of fake account:", prior_fake)
print("Likelihood of high activity:", likelihood)
print("Evidence:", round(evidence, 3))
print("Probability of fake account:", round(posterior, 4))
print("Percentage:", round(posterior * 100, 2), "%")
```
