```python
#Task:1
# Function to calculate probability

def calculate_probability(event_count, total_outcomes):
    probability = event_count / total_outcomes
    return probability

# Tossing a coin
heads = 1
total_outcomes = 2

result = calculate_probability(heads, total_outcomes)

print("Probability of getting Heads:", result)
print("Percentage:", result * 100, "%")
```

```python
# Task:2
import random

count_six = 0

# Roll the dice 100 times
for i in range(100):
    dice = random.randint(1, 6)

    if dice == 6:
        count_six += 1

# Calculate experimental probability
probability = count_six / 100

print("Total rolls: 100")
print("Number of times 6 appeared:", count_six)
print("Experimental probability:", probability)
```

```python
# Task:3
# Last 20 songs played on Spotify
songs = [
    "pop", "rock", "pop", "hiphop", "jazz",
    "pop", "rock", "pop", "jazz", "hiphop",
    "pop", "rock", "pop", "jazz", "pop",
    "rock", "hiphop", "pop", "jazz", "pop"
]

# Count pop songs
pop_count = songs.count("pop")

# Calculate probability
probability = pop_count / len(songs)

print("Total songs:", len(songs))
print("Number of Pop songs:", pop_count)
print("Probability of selecting a Pop song:", probability)
```

```python
# Task:4
# Users who ordered food
food_users = [1, 2, 3, 4, 5, 6, 7, 8]

# Users who ordered dessert
dessert_users = [2, 3, 5, 7, 9]

# Find users who ordered both
both_users = set(food_users) & set(dessert_users)

# Find total unique users
all_users = set(food_users) | set(dessert_users)

# Calculate probability
probability = len(both_users) / len(all_users)

print("Users who ordered both:", sorted(both_users))
print("Total unique users:", len(all_users))
print("Probability of ordering both:", round(probability, 2))
```

```python
# Task:5
# Users who bought a phone
phone_users = 100

# Users who bought both a phone and headphones
both_bought = 60

# Calculate conditional probability
probability = both_bought / phone_users

print("Users who bought a phone:", phone_users)
print("Users who bought both phone and headphones:", both_bought)
print("Probability of buying headphones given phone purchase:", probability)
print("Percentage:", probability * 100, "%")

if probability > 0.5:
    print("Recommend headphones to phone buyers.")
else:
    print("Show other product recommendations.")
```
