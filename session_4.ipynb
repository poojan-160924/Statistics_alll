```python
# Task:1

import matplotlib.pyplot as plt

# Daily order counts for 10 days
orders = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]

# Display the order counts
print("Daily order counts:", orders)

# Create a histogram
plt.hist(orders, bins=10, edgecolor="black")

plt.title("Daily Swiggy Order Counts")
plt.xlabel("Number of Orders")
plt.ylabel("Number of Days")

plt.show()

print("The data is right skewed.")
```

```python
# Task:2
import statistics

followers = [5, 7, 8, 8, 9, 10, 12, 15, 95]

# Calculate mean and median
mean_followers = statistics.mean(followers)
median_followers = statistics.median(followers)

print("Follower counts:", followers)
print("Mean:", round(mean_followers, 2))
print("Median:", median_followers)

if mean_followers > median_followers:
    print("The data is right skewed.")
else:
    print("The data is left skewed.")
```

```python
# Task:3
import statistics

order_amounts = [150, 200, 250, 180, 300, 220, 500, 2500]

mean_amount = statistics.mean(order_amounts)
median_amount = statistics.median(order_amounts)

print("Order amounts:", order_amounts)
print("Mean order amount:", mean_amount)
print("Median order amount:", median_amount)

if mean_amount > median_amount:
    print("The data is right skewed.")
```

```python
# Task:4
import matplotlib.pyplot as plt
from collections import Counter

# Flipkart product ratings
ratings = [3, 3, 4, 4, 4, 5, 5, 5, 5, 5]

# Count each rating
rating_count = Counter(ratings)

print("Product ratings:", ratings)
print("Rating frequency:", dict(rating_count))

# Prepare graph data
rating_values = list(rating_count.keys())
frequencies = list(rating_count.values())

# Draw bar graph
plt.bar(rating_values, frequencies, edgecolor="black")

plt.title("Flipkart Product Ratings")
plt.xlabel("Ratings")
plt.ylabel("Number of Customers")

plt.xticks([1, 2, 3, 4, 5])

plt.show()

print("The distribution is left skewed.")
```
