```python
#Task:1
# Duration of 5 songs in seconds

song_durations = [210, 180, 245, 200, 175]

# Calculate total playlist duration
total_duration = sum(song_durations)

print("Song durations:", song_durations)
print("Total playlist duration:", total_duration, "seconds")
```

```python
# Task:2
# Ratings given by 3 friends to 3 food items

ratings = [
    [5, 4, 3],
    [4, 5, 2],
    [3, 4, 5]
]

# Display the rating matrix
print("Food Rating Matrix:")

for row in ratings:
    print(row)

# Display ratings for the second food item
print("\nRatings for the second food item:")

for i in range(len(ratings)):
    print("Friend", i + 1, ":", ratings[i][1])
```

```python
# Task:3
# Orders of 4 dishes across 3 days

orders = [
    [10, 15, 20, 12],
    [12, 18, 16, 14],
    [15, 20, 25, 10]
]

# Calculate total orders for each dish
total_dishes = len(orders[0])

print("Daily order details:")

for row in orders:
    print(row)

print("\nTotal orders for each dish:")

for i in range(total_dishes):
    total = 0

    for day in orders:
        total += day[i]

    print("Dish", i + 1, ":", total, "orders")
```

```python
# Task:4
# Product details: [Price, Quantity Sold]

products = [
    [500, 10],
    [1200, 5],
    [800, 8],
    [1500, 4],
    [300, 15]
]

# Function to calculate product revenue
def calculate_revenue(products):

    for i in range(len(products)):
        price = products[i][0]
        quantity = products[i][1]

        revenue = price * quantity

        print("Product", i + 1)
        print("Price:", price)
        print("Quantity sold:", quantity)
        print("Total revenue:", revenue)
        print()


# Display product revenue
calculate_revenue(products)
 
```
